```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
from selenium.webdriver.chrome.options import Options
```

```python
chrome_options = Options()

navegador = webdriver.Chrome(options=chrome_options)

navegador.maximize_window()

navegador.get("https://monthly-sales-wizard.lovable.app/")

time.sleep(3)

navegador.find_element(By.XPATH, '//*[@id="email"]').send_keys('iago@teste')

time.sleep(2)

navegador.find_element(By.XPATH, '//*[@id="password"]').send_keys('Teste1234')

time.sleep(2)

navegador.find_element(By.XPATH, '//*[@id="root"]/div[2]/div/div/div[2]/form/button').click()
```
